# Frame Interpolator Working Demo

```python
import io
import ipywidgets as widgets
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from IPython.display import display
from PIL import Image

import mlutils 
import fiutils
from model import FrameInterpolator
```

```python
device = mlutils.get_device()
print(device)
```

### Import my Model build

```python
model = FrameInterpolator().to(device)
```

### Loading the trained model

```python
weights_path = "best_frame_interpolator.pth"

try:
    model.load_state_dict(torch.load(weights_path, map_location=device))
    print("Model loaded successfully!")
except FileNotFoundError:
    print(f"Error: Could not find {weights_path}.")
```

### Create Buttons

```python
upload_1 = widgets.FileUpload(accept="image/*", multiple=False, description="Upload Start Frame")
upload_2 = widgets.FileUpload(accept="image/*", multiple=False, description="Upload End Frame")
btn_run = widgets.Button(description="Generate In-Between", button_style="success")
out = widgets.Output()
```

## The Demo

```python
target_size = (448, 256)

btn_run.on_click(lambda b: fiutils.generate_demo(out, upload_1, upload_2, target_size, device, model))
display(widgets.HBox([upload_1, upload_2]))
display(btn_run)
display(out)
```

### GIF-Maker
Run this code block to see your output in action!

```python
img1 = fiutils.process_widget(upload_1, target_size, device)
img3 = fiutils.process_widget(upload_2, target_size, device)

if img1 is not None and img3 is not None:
    anim = fiutils.create_gif(model, device, img1=img1, img3=img3)
    display(anim)
else:
    print("Please upload images first!")
```
